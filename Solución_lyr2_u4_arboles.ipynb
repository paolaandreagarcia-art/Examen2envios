```python
def LCA(x, p, q):

    if not x or x[0] is None:
        return None

    i = 0

    while i < len(x) and x[i] is not None:

        actual = x[i]

        if p < actual and q < actual:
            i = 2 * i + 1

        elif p > actual and q > actual:
            i = 2 * i + 2

        else:
            return actual

    return None
```

4.1. ¿Por qué los árboles binarios son más eficientes que las listas en ciertas operaciones?
**Respuesta:** Los árboles binarios pueden ser más eficientes que las listas porque organizan los datos mediante relaciones jerárquicas entre nodos, permitiendo reducir la cantidad de elementos que deben revisarse durante determinadas operaciones.

En una lista, para encontrar un elemento específico generalmente es necesario recorrer los elementos uno por uno hasta encontrarlo.

En cambio, en un árbol binario de búsqueda, los elementos se organizan de manera que los valores menores se encuentran a la izquierda y los mayores a la derecha.

Esto permite descartar una parte del árbol en cada comparación.

4.2. ¿Cómo influye la altura de un árbol binario en su eficiencia en términos de búsqueda y acceso a datos?
**Respuesta:** La altura de un árbol binario representa la longitud del camino más largo desde la raíz hasta una hoja.

Esta característica influye directamente en el tiempo necesario para buscar un elemento.

Cuando un árbol está balanceado, su altura es pequeña y permite encontrar los elementos realizando menos comparaciones.

Por el contrario, cuando un árbol está desbalanceado, puede convertirse prácticamente en una estructura lineal, aumentando el número de operaciones necesarias.
4.3 ¿Cómo se implementa la eliminación de un nodo en un árbol binario de búsqueda y por qué puede afectar la estructura del árbol?
**Respuesta:** La eliminación de un nodo en un árbol binario de búsqueda depende de la cantidad de hijos que tenga el nodo que se desea eliminar.

Existen tres situaciones:

**Caso 1.** El nodo no tiene hijos

Se elimina directamente, porque no tiene otros nodos que dependan de él.

**Caso 2.** El nodo tiene un hijo

Se elimina el nodo y su padre queda conectado directamente con el hijo del nodo eliminado.

**Caso 3.** El nodo tiene dos hijos

Se busca el sucesor inorden, que corresponde al menor valor del subárbol derecho, o el predecesor inorden, que corresponde al mayor valor del subárbol izquierdo.

Se reemplaza el valor del nodo que se desea eliminar y posteriormente se elimina el nodo utilizado como reemplazo.

```python
class Nodo:
    def __init__(self,valor):
        self.valor=valor
        self.izquierdo=None
        self.derecho=None

def insertar(raiz,valor):
    if raiz is None: return Nodo(valor)
    if valor<raiz.valor: raiz.izquierdo=insertar(raiz.izquierdo,valor)
    elif valor>raiz.valor: raiz.derecho=insertar(raiz.derecho,valor)
    return raiz

def altura(raiz):
    if raiz is None: return -1
    return 1+max(altura(raiz.izquierdo),altura(raiz.derecho))

def verificar_balance(raiz):
    if raiz is None: return True,-1
    bi,hi=verificar_balance(raiz.izquierdo)
    bd,hd=verificar_balance(raiz.derecho)
    return bi and bd and abs(hi-hd)<=1,1+max(hi,hd)

def inorden(raiz):
    if raiz:
        inorden(raiz.izquierdo)
        print(raiz.valor,end=" ")
        inorden(raiz.derecho)

def construir_balanceado(datos,inicio,fin,movimientos,padre=None,lado="raíz"):
    if inicio>fin: return None
    medio=(inicio+fin)//2
    valor=datos[medio]
    movimientos.append("Ubicar nodo "+str(valor)+" como "+lado+(" de "+str(padre) if padre is not None else ""))
    raiz=Nodo(valor)
    raiz.izquierdo=construir_balanceado(datos,inicio,medio-1,movimientos,valor,"izquierdo")
    raiz.derecho=construir_balanceado(datos,medio+1,fin,movimientos,valor,"derecho")
    return raiz

def balancear(raiz):
    datos=[]
    def guardar(nodo):
        if nodo:
            guardar(nodo.izquierdo)
            datos.append(nodo.valor)
            guardar(nodo.derecho)
    guardar(raiz)
    movimientos=[]
    nuevo=construir_balanceado(datos,0,len(datos)-1,movimientos)
    return nuevo,movimientos

# Programa principal
datos=list(range(1,28))
raiz=None
for valor in datos: raiz=insertar(raiz,valor)

print("ARBOL ORIGINAL")
print("Cantidad de nodos:",len(datos))
print("Altura:",altura(raiz))
print("¿Está balanceado?:",verificar_balance(raiz)[0])

nuevo_arbol,movimientos=balancear(raiz)

print("\nMOVIMIENTOS PARA BALANCEAR")
for movimiento in movimientos: print(movimiento)

print("\nARBOL BALANCEADO")
print("Recorrido inorden:")
inorden(nuevo_arbol)
print("\nAltura:",altura(nuevo_arbol))
print("¿Está balanceado?:",verificar_balance(nuevo_arbol)[0])
```
